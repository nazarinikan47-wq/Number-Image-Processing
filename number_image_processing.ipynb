# Preprocess

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(digits.data , digits.target , test_size=0.3)
x_train[0]
```

# Normalize

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,1))
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train[0]
```

# PCA

```python
x_train.shape , x_test.shape
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=32)

x_train = pca.fit_transform(x_train)
x_test = pca.transform(x_test)
```

```python
x_train.shape , x_test.shape
```

# performance metric

```python
from sklearn.metrics import accuracy_score , recall_score , precision_score

def calculate_metrics(y_train,y_test,y_pred_train,y_pred_test):
    acc_train = accuracy_score(y_true=y_train,y_pred=y_pred_train)
    acc_test = accuracy_score(y_true=y_test,y_pred=y_pred_test)
    
    p = precision_score(y_true=y_test,y_pred=y_pred_test , average='weighted')
    r = recall_score(y_true=y_test,y_pred=y_pred_test , average='weighted')

    print(f'acc train : {acc_train} and acc test : {acc_test} \n and precision : {p} and recall : {r}')
    return acc_train , acc_test , p , r
```

# classfication

## 1 - Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(max_depth=128, n_estimators=256)
rf.fit(x_train,y_train)
```

```python
y_pred_train = rf.predict(x_train)
y_pred_test = rf.predict(x_test)

acc_train_rf , acc_test_rf ,p_rf ,r_rf = calculate_metrics(y_train,y_test,y_pred_train,y_pred_test)  
```

## 2 - SVM

```python
from sklearn.svm import SVC

svm = SVC(kernel='poly' , probability=True)
svm.fit(x_train,y_train)
```

```python
y_pred_train = svm.predict(x_train)
y_pred_test = svm.predict(x_test)

acc_train_svm , acc_test_svm ,p_svm ,r_svm = calculate_metrics(y_train,y_test,y_pred_train,y_pred_test)  
```

## 3 - ANN

```python
from sklearn.neural_network import MLPClassifier

ann = MLPClassifier(hidden_layer_sizes=256)
ann.fit(x_train,y_train)
```

```python
y_pred_train = ann.predict(x_train)
y_pred_test = ann.predict(x_test)

acc_train_ann , acc_test_ann ,p_ann ,r_ann = calculate_metrics(y_train,y_test,y_pred_train,y_pred_test)  
```

## 4 - KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(x_train,y_train)
```

```python
y_pred_train = knn.predict(x_train)
y_pred_test = knn.predict(x_test)

acc_train_knn , acc_test_knn ,p_knn ,r_knn = calculate_metrics(y_train,y_test,y_pred_train,y_pred_test)  
```

# Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import os
```

```python
def plot_confusion_matrix(model, model_name):

    y_pred = model.predict(x_test)

    cm = confusion_matrix(y_test, y_pred)

    fig, ax = plt.subplots(figsize=(6,6))

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm
    )

    disp.plot(cmap="Blues", ax=ax)

    plt.title(f"{model_name} Confusion Matrix")

    plt.savefig(
        f"images/{model_name.lower().replace(' ','_')}_confusion_matrix.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
plot_confusion_matrix(rf, "Random Forest")
plot_confusion_matrix(svm, "SVM")
plot_confusion_matrix(ann, "Neural Network")
plot_confusion_matrix(knn, "KNN")
```

# Multi-class ROC

```python
from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc
from sklearn.multiclass import OneVsRestClassifier
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_multiclass_roc(model, model_name):

    classifier = OneVsRestClassifier(model)
    classifier.fit(x_train, y_train)

    y_score = classifier.predict_proba(x_test)

    classes = np.unique(y_train)
    y_test_bin = label_binarize(y_test, classes=classes)

    plt.figure(figsize=(8,6))

    for i in range(len(classes)):
        fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
        roc_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"Class {i} (AUC = {roc_auc:.2f})"
        )

    plt.plot([0,1],[0,1],'k--')

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")

    plt.title(f"{model_name} ROC Curve")

    plt.legend(loc="lower right", fontsize=8)

    plt.savefig(
        f"images/{model_name.lower().replace(' ','_')}_roc_curve.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
plot_multiclass_roc(rf,"Random Forest")
plot_multiclass_roc(svm,"SVM")
plot_multiclass_roc(ann,"Neural Network")
plot_multiclass_roc(knn,"KNN")
```
